```python
import sys
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent
sys.path.append(str(ROOT))

df = pd.read_parquet(ROOT / "data" / "silver" / "matches.parquet")
df.shape
```

```python
print("phase nulls:", df["phase"].isna().sum())
print(df.groupby(["league", "season", "phase"]).size())
df["result"].value_counts() 
```

```python
key_cols = ["match_id", "league", "season", "start_time", "phase",
            "home_team_id", "away_team_id", "home_goals", "away_goals", "result"]

print("rows:", len(df))
print("duplicate match_ids:", df["match_id"].duplicated().sum())
print("nulls in key columns:", df[key_cols].isna().sum().sum())

expected = df.apply(lambda r: "H" if r.home_goals > r.away_goals
                    else "A" if r.home_goals < r.away_goals else "D", axis=1)
print("result mismatches:", (expected != df["result"]).sum())

print("\npossession nulls per league/season:")
print(df[df["home_ball_possession"].isna()].groupby(["league", "season"]).size())
```
